```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer

def clean_categories(data):
    data = pd.DataFrame(data).copy()

    for col in data.columns:
        data[col] = data[col].map(
            lambda value: str(value) if pd.notna(value) else np.nan
        )

    return data

categorical_transformer = Pipeline(
    steps=[
        (
            "clean",
            FunctionTransformer(clean_categories)
        ),
        (
            "imputer",
            SimpleImputer(
                strategy="constant",
                fill_value="Missing"
            )
        ),
        (
            "encoder",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.linear_model import LinearRegression

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)
```

```python
model.fit(X_train, y_train)

print("Model training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = model.predict(X_test)

print("MAE:", round(mean_absolute_error(y_test, y_pred), 2))
print("RMSE:", round(mean_squared_error(y_test, y_pred) ** 0.5, 2))
print("R2:", round(r2_score(y_test, y_pred), 4))
```

```python
param_distributions = {
    "regressor__n_estimators": randint(100, 501),
    "regressor__max_depth": randint(3, 10),
    "regressor__learning_rate": uniform(0.01, 0.19),
    "regressor__subsample": uniform(0.6, 0.4),
    "regressor__colsample_bytree": uniform(0.6, 0.4),
    "regressor__min_child_weight": randint(1, 8)
}
```

```python
from xgboost import XGBRegressor
from sklearn.pipeline import Pipeline

tuning_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            XGBRegressor(
                objective="reg:squarederror",
                random_state=42,
                n_jobs=1
            )
        )
    ]
)
```

```python
random_search = RandomizedSearchCV(
    estimator=tuning_pipeline,
    param_distributions=param_distributions,
    n_iter=10,
    scoring="neg_mean_absolute_error",
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=-1
)
```

```python
random_search.fit(X_train, y_train)

print("Hyperparameter tuning completed!")
```

```python
print("Best parameters:")

for parameter, value in random_search.best_params_.items():
    print(parameter, ":", value)

print(
    "\nBest cross-validation MAE:",
    round(-random_search.best_score_, 2)
)
```
